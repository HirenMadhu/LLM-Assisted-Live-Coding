```python
import os, glob
from datasets import Dataset, concatenate_datasets, Features, Value, Sequence

root = "sonicpi_embeddings_full.arrow"  # folder containing dataset_info.json and data/
shards = ["sonicpi_embeddings_full.arrow/data-00000-of-00001.arrow"]
parts = [Dataset.from_file(p) for p in shards]
ds = parts[0] if len(parts) == 1 else concatenate_datasets(parts)

# Cast to an explicit schema equivalent to your JSON
features = Features({
    "code": Value("string"),
    "code_embedding": Sequence(Value("float32"), length=768),
    "wav_embedding": Sequence(Value("float32"), length=768),
})
ds = ds.cast(features)

# (Optional) re-save so future load_from_disk works without hacks
ds.save_to_disk("sonicpi_embeddings_full_new.arrow")
```

```python
from datasets import load_from_disk
import torch
import numpy as np
ds = load_from_disk("s2-drum/10-cand/label_embs_s2-drum-10-cand.arrow")
code, wav = [], []
for ex in ds:
    code.append(ex["code_embedding"])
    wav.append(ex["wav_embedding"])
code = torch.tensor(np.asarray(code), dtype=torch.float32)
wav = torch.tensor(np.asarray(wav), dtype=torch.float32)
```
